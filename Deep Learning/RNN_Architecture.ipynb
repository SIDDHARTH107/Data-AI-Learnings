```python
# Importing dependencies
from keras import Sequential
from keras.layers import Dense, SimpleRNN # Importing 2 layers
```

```python
model = Sequential()

model.add(SimpleRNN(3, input_shape=(4, 5)))
model.add(Dense(1, activation='relu'))

model.summary()
```

```python
model = Sequential()

# Adding some layers
# 1st one is a simple RNN layers with 3 nodes and another one is a dense layer with 1 node
model.add(SimpleRNN(3, input_shape=(4, 5)))
model.add(Dense(1, activation='sigmoid')) # Sigmoid cuz here it is a binary classification task

model.summary()
```

```python
model = Sequential()

model.add(SimpleRNN(3, input_shape=(4, 5)))
model.add(Dense(1, activation='tanh'))

model.summary()
```

```python
# Let's check all the trainable parameters
# This is the input matrix
print(model.get_weights()[0].shape)
print(model.get_weights()[0])
```

```python
# This is the feedback conncetion which is a 3 X 3 matrix
print(model.get_weights()[1].shape)
print(model.get_weights()[1])
```

```python
# Let's check all the trainable parameters
# These are 3 biases
print(model.get_weights()[2].shape)
model.get_weights()[2]
```

```python
# One final bias
print(model.get_weights()[3].shape)
model.get_weights()[3]
```
